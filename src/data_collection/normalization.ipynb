```python
import pandas as pd

df = pd.read_csv("../../data/guide_box.csv")
df
```

```python
# 기존 해상도 정보
initial_width, initial_height, = 640, 480

# 비율 계산(0.0~1.0)
df["x1_ratio"] = df["x1"] / initial_width
df["y1_ratio"] = df["y1"] / initial_height
df["x2_ratio"] = df["x2"] / initial_width
df["y2_ratio"] = df["y2"] / initial_height
```

```python
new_df = df.drop(["x1", "y1", "x2", "y2"], axis=1)
new_df
new_df.to_csv("../../data/guide_box_ratio.csv", index=False)
```
